```python
# ==============================================================
# Experiment: Reading and Writing CSV, Excel, and JSON Files
# Aim:
# 1. Read data from a CSV file.
# 2. Save the data into Excel and JSON formats.
# 3. Read the saved files back.
# 4. Add a new column.
# 5. Save the modified data.
# ==============================================================


# --------------------------------------------------------------
# Import Library
# --------------------------------------------------------------

# pandas is a Python library used for:
# • Reading and writing data files
# • Data cleaning
# • Data analysis
# • Data manipulation
#
# 'pd' is the short name (alias) for pandas.

import pandas as pd


# --------------------------------------------------------------
# Step 1: Read CSV File from Internet
# --------------------------------------------------------------

# Store the URL of the CSV file in a string variable.
# String is an immutable data type.

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"

# read_csv()
# Reads a CSV (Comma Separated Values) file
# and stores it as a DataFrame.

df_csv = pd.read_csv(url)

# Print a heading.
print("----- CSV Data -----")

# shape
# Returns the number of rows and columns.
# Output format: (rows, columns)

print("Shape:", df_csv.shape)
```

```python
# --------------------------------------------------------------
# Step 2: Save Data into Excel File
# --------------------------------------------------------------

# to_excel()
# Saves the DataFrame as an Excel (.xlsx) file.

# index=False
# Prevents pandas from saving row numbers.

# sheet_name
# Specifies the worksheet name.

df_csv.to_excel(
    "tips_output.xlsx",
    index=False,
    sheet_name="Tips"
)

# read_excel()
# Reads an Excel file into a DataFrame.

df_excel = pd.read_excel(
    "tips_output.xlsx",
    sheet_name="Tips"
)

print("\n----- Excel Round-Trip -----")

print("Shape after Excel:", df_excel.shape)

```

```python
# --------------------------------------------------------------
# Step 3: Save Data into JSON File
# --------------------------------------------------------------

# JSON = JavaScript Object Notation
# It stores data in key-value format.

# to_json()
# Saves the DataFrame into JSON format.

# orient="records"
# Saves each row as one JSON object.

# indent=2
# Makes the JSON file easy to read.

df_csv.to_json(
    "tips_output.json",
    orient="records",
    indent=2
)

# read_json()
# Reads a JSON file.

df_json = pd.read_json("tips_output.json")

print("\n----- JSON Round-Trip -----")

print("Shape after JSON:", df_json.shape)

```

```python
# --------------------------------------------------------------
# Step 4: Create a New Column
# --------------------------------------------------------------

# Add a new column named "tip_percentage"

# Formula:
# (Tip / Total Bill) × 100

# round()
# Rounds decimal values.
# round(value,2) keeps only two decimal places.

df_csv["tip_percentage"] = round(
    (df_csv["tip"] / df_csv["total_bill"]) * 100,
    2
)

print("\n----- Data After Adding New Column -----")

print(df_csv.head(3))
```

```python
# --------------------------------------------------------------
# Step 5: Save Modified Data into CSV File
# --------------------------------------------------------------

# to_csv()
# Saves the DataFrame into a CSV file.

df_csv.to_csv(
    "tips_modified.csv",
    index=False
)

print("\nModified CSV file saved successfully!")
```
