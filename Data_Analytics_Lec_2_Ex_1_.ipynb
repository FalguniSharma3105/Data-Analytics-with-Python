```python
# ==========================================================
# Experiment: Sales Data Analysis using Python Lists
# Aim:
# 1. Store sales data of different regions.
# 2. Calculate total sales.
# 3. Calculate average sales.
# 4. Find the region with the lowest sales.
# ==========================================================


# ----------------------------------------------------------
# Step 1: Create two lists
# ----------------------------------------------------------

# A list is a mutable data type.
# It stores multiple values in a single variable.
# Lists are created using square brackets [ ].

sales = [420, 455, 610, 300, 175]

# This list stores the names of the regions.

regions = ["North", "South", "East", "West", "Central"]

```

```python
# ----------------------------------------------------------
# Step 2: Calculate Total Sales
# ----------------------------------------------------------

# sum()
# Adds all numbers present in the list.

total = sum(sales)

print(total)

# READ THE OUTPUT BELOW
```

```python
# ----------------------------------------------------------
# Step 3: Calculate Average Sales
# ----------------------------------------------------------

# len()
# Returns the total number of elements in the list.

# Formula:
# Average = Total Sales / Number of Regions

average = total / len(sales)

print(average)

# READ THE OUTPUT BELOW
```

```python
# ----------------------------------------------------------
# Step 4: Find the Lowest Sales Value
# ----------------------------------------------------------

# min()
# Returns the smallest value in the list.

lowest_sales = min(sales)

print(lowest_sales)

# READ THE OUTPUT BELOW
```

```python
# ----------------------------------------------------------
# Step 5: Find the Position (Index) of the Lowest Sales
# ----------------------------------------------------------

# index()
# Returns the position of the specified value.

lowest_index = sales.index(lowest_sales)

print(lowest_index)

# READ THE OUTPUT BELOW
```

```python
# ----------------------------------------------------------
# Step 6: Find the Region Name
# ----------------------------------------------------------

# Use the index obtained above to access the
# corresponding region from the regions list.

lowest_region = regions[lowest_index]

print(lowest_region)

# READ THE OUTPUT BELOW
```

```python
# ----------------------------------------------------------
# Step 7: Display All Results Together
# ----------------------------------------------------------

# print()
# Displays output on the screen.
# \n works like Enter; use it to leave one line.

print("Sales Data:", sales)

print("\n Regions:", regions)

print("\n Total Sales:", total)

print("\n Average Sales:", average)

print("\n Lowest Sales:", lowest_sales)

print("\n Lowest Performing Region:", lowest_region)
```
