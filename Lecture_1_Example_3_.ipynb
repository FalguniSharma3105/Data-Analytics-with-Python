```python
# ==========================================================
# EXAMPLE 3: String Operations in Python
# Aim: Count the number of words in a sentence and
#      check whether a specific word is present.
# ==========================================================

# Step 1: Store a review in a string variable.
# A string is a sequence of characters enclosed in quotes.

review = "The product quality was excellent, but delivery was delayed by two days."

# Step 2: Split the string into individual words.
# The split() function divides the sentence into a list of words.
# By default, it splits wherever there is a space.

words = review.split()

# Step 3: Count the number of words.
# The len() function returns the total number of items in the list.

print("Word count:", len(words))

# Step 4: Check whether the word "delayed" is present.
# The 'in' operator searches for a word or text inside a string.
# If found, it returns True; otherwise, it returns False.

print("Contains 'delayed':", "delayed" in review)
```
