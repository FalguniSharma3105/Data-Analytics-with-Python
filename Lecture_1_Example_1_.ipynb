```python
# ==========================================================
# Experiment: Dictionary Data Type in Python
# Aim: Create a dictionary, display it, and check its data type.
# ==========================================================

# Step 1: Create a dictionary named 'student'.
# A dictionary stores data in the form of Key : Value pairs.
# Keys are unique and enclosed in quotes.
# Values can be of different data types.

student = {
    "Name": "Amit",
    "Age": 21,
    "Marks": 85
}

# Step 2: Display a heading.
# The print() function is used to display output on the screen.
print("Student Details:")

# Step 3: Print the complete dictionary.
# Python displays all key-value pairs inside curly braces {}.
print(student)

# Step 4: Leave one blank line for better output formatting.
# '\n' means New Line.
print("\nData Type of student:")

# Step 5: Check the data type of the variable.
# The type() function returns the data type of any variable.
# Since 'student' is a dictionary, the output will be:
# <class 'dict'>
print(type(student))
```
