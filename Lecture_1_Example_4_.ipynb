```python
# ===============================================================
# Example 4: Types of Data in Python
# Aim: Store structured, semi-structured, and unstructured data
#      and display the data type of each.
# ===============================================================

# Step 1: Create a list named 'samples'.
# Each item in the list is a dictionary.
# Every dictionary contains:
#   1. 'type'  -> the category of data
#   2. 'value' -> the actual data

samples = [

    {
        "type": "structured",
        "value": {
            "Name": "Raj",
            "Age": 20
        }
    },

    {
        "type": "semi-structured",
        "value": {
            "tags": ["python", "data"]
        }
    },

    {
        "type": "unstructured",
        "value": "A long free-text customer comment..."
    }

]

# Step 2: Use a for loop to visit each item in the list.
for s in samples:

    # Step 3:
    # s["type"] gives the category of data.
    # s["value"] gives the actual data.
    # type() returns the data type.
    # __name__ returns only the name of the data type.
    print(f"{s['type']}: {type(s['value']).__name__}")
```
